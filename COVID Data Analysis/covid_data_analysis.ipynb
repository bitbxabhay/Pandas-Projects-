```python
#recovery rate 
df["Recovery Rate"] = (df["Recovered"]/df["Confirmed"])*100
df.sort_values("Recovery Rate",ascending=False)
```

```python
#death rate 
df["Death Rate"] = (df["Deaths"]/df["Confirmed"])*100
df.sort_values("Death Rate",ascending=False)
```

```python
#active case rate
df["Active Case Rate"] = (df["Active"]/df["Confirmed"])*100
df.sort_values("Active Case Rate",ascending=False)
```

```python
#mean confirmed cases
print(f"Confirmed Cases Mean : {df["Confirmed"].mean()}")
```
