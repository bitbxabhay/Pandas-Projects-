# Sales Dashboard Analysis

```python
#importing pandas and numpy 
import pandas as pd 
import numpy as np 
```

```python
#Dataframe visualization 
df = pd.read_csv("superstore.csv", encoding="latin1")
```

```python
df.head(2)
```

```python
df.shape
```

```python
#total sales
df["Sales"].sum()
```

```python
#total profit
df["Profit"].sum()
```

```python
#total order
df["Order ID"].nunique()
```

```python
#avg order value
df["Sales"].mean()
```

```python
#top 10 products by sales
top_prod = df.groupby("Product Name")["Sales"].sum()
top_prod.sort_values(ascending=False).head(10)
```

```python
#top 10 products by profit
top_prof = df.groupby("Product Name")["Profit"].sum()
top_prof.sort_values(ascending = False).head(10)
```

```python
#least profitable products
least_prof = df.groupby("Product Name")["Profit"].sum()
least_prof.sort_values(ascending = True ).head(10)
```

```python
#sales by category
sales_by_categ = df.groupby("Category")["Sales"].sum()
sales_by_categ.sort_values(ascending=False).head()
```

```python
#profit by category
prof_by_categ = df.groupby("Category")["Profit"].sum()
prof_by_categ.sort_values(ascending = False).head()
```

```python
#sales by sub-category
sales_by_sub_category = df.groupby("Sub-Category")["Sales"].sum()
sales_by_sub_category.sort_values(ascending = False ).head()
```

```python
#profit by sub-category
profit_by_sub_category = df.groupby("Sub-Category")["Profit"].sum()
profit_by_sub_category.sort_values(ascending = False ).head()
```

```python
#sales by region 
sales_by_region = df.groupby("Region")["Sales"].sum()
sales_by_region.sort_values(ascending = False ).head()
```

```python
#profit by region
profit_by_region = df.groupby("Region")["Profit"].sum()
profit_by_region.sort_values(ascending = False ).head()
```

```python
#top 10 customers by sales 
top_customers_by_sales = df.groupby("Customer Name")["Sales"].sum()
top_customers_by_sales.sort_values(ascending = False ).head(10)
```

```python
#top 10 customers by profit 
top_customers_by_profit = df.groupby("Customer Name")["Profit"].sum()
top_customers_by_profit.sort_values(ascending = False ).head(10)
```

```python
#most ordered products
most_ordered_products = df.groupby("Product Name")["Quantity"].sum()
most_ordered_products.sort_values(ascending = False).head()
```

```python
#monthly sales trend
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales
```

```python
#montly profit trend
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_profit = df.groupby(df["Order Date"].dt.to_period("M"))["Profit"].sum()
monthly_profit
```

```python
#year wise sale 
yearwise_Sale = df.groupby(df["Order Date"].dt.year)["Sales"].sum()
yearwise_Sale
```

```python
#year wise profit 
yearwise_profit = df.groupby(df["Order Date"].dt.year)["Profit"].sum()
yearwise_profit
```

```python
#segment wise sale 
segment_wise_sale = df.groupby("Segment")["Sales"].sum()
segment_wise_sale.sort_values(ascending=False)
```

```python
#segment wise profit 
segment_wise_profit = df.groupby("Segment")["Profit"].sum()
segment_wise_profit.sort_values(ascending = False )
```

```python
#most profitable category in each region
most_profitable_category_in_each_region = df.groupby(["Region","Category"])["Profit"].sum()
most_profitable_category_in_each_region
```

```python
#avg profit per order
avg_profit = df["Profit"].mean()
avg_profit
```

```python
#product making loss 
product_making_loss = df.groupby("Product Name")["Profit"].sum()
product_making_loss[product_making_loss < 0]
```

```python
#city with highest sales
city_with_highest_sales = df.groupby("City")["Sales"].sum()
city_with_highest_sales.sort_values(ascending = False).head(20)
```

```python
#city with highest profit 
city_with_highest_profit = df.groupby("City")["Profit"].sum()
city_with_highest_profit.sort_values(ascending = False).head(20)
```
